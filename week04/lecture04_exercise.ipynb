## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2000 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
import plotly.express as px

df = px.data.gapminder()

continent_choice = 'Asia'
df_filtered = df.loc[
    (df['continent'] == continent_choice) & 
    (df['year'].isin([2002, 2007]))
]

fig = px.scatter(
    df_filtered, 
    x="gdpPercap", 
    y="lifeExp",
    color="year",
    hover_name="country",
    log_x=True,
    title=f"GDP vs Life Expectancy in {continent_choice} (Comparison: 2002 vs 2007)",
    color_discrete_map={
        2002: "red",
        2007: "blue"
    },
    labels={
        "gdpPercap": "GDP per Capita (USD)",
        "lifeExp": "Life Expectancy (Years)",
        "year": "Year"
    }
)

fig.update_traces(marker=dict(size=12))
fig.update_layout(
    plot_bgcolor='white',
    xaxis=dict(showgrid=True, gridcolor='LightGray'),
    yaxis=dict(showgrid=True, gridcolor='LightGray')
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
import plotly.express as px
import pandas as pd

df = px.data.gapminder()

df_2007 = df.loc[df['year'] == 2007].copy()

highlight_continent = 'Europe'
df_2007['focus_group'] = df_2007['continent'].apply(
    lambda x: highlight_continent if x == highlight_continent else 'Other'
)

fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="focus_group",
    hover_name="country",
    log_x=True,
    size_max=60,
    title=f"The Global Story of {highlight_continent} (2007)",
    color_discrete_map={
        highlight_continent: "#CB5B9B",
        "Other": "#E5ECF6"
    },
    labels={
        "gdpPercap": "GDP per Capita (USD)",
        "lifeExp": "Life Expectancy (Years)"
    }
)


fig.add_annotation(
    x=4000, y=70,
    text="<b>The Rise of Asian Giants</b><br>China and India represent nearly 40% of the population.<br>Their movement toward higher life expectancy<br>drives the global average upward.",
    showarrow=True,
    arrowhead=2,
    ax=50, ay=-120,
    bgcolor="white",
    bordercolor="#636EFA",
    borderwidth=2,
    font=dict(size=12)
)

fig.update_layout(
    plot_bgcolor='white',
    xaxis=dict(showgrid=True, gridcolor='LightGrey'),
    yaxis=dict(showgrid=True, gridcolor='LightGrey'),
    showlegend=False
)

# 7. Show output
fig.show()
```
